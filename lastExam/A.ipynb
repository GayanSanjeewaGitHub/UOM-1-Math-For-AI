confidenace interval

```python
import numpy as np
from scipy import stats

# Given data
sample_mean = 2.62
sample_std = 0.42
sample_size = 150
confidence_level = 0.95

# Calculate the standard error
standard_error = sample_std / np.sqrt(sample_size)

# Calculate the critical value (z-score for 95% confidence)
z_critical = stats.norm.ppf((1 + confidence_level) / 2)

# Calculate the margin of error
margin_of_error = z_critical * standard_error

# Calculate the confidence interval
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

print(f"The 95% confidence interval for the mean GPA is: {confidence_interval}")
```

inverse matrics


```python
import numpy as np

# Define your 3x3 matrix
matrix = np.array([
    [1, 2, 3],
    [4, 6, 6],
    [7, 8, 9]
])

# Check if the matrix is invertible (determinant != 0)
determinant = np.linalg.det(matrix)
if determinant == 0:
    print("The matrix is singular and cannot be inverted.")
else:
    # Calculate the inverse
    inverse_matrix = np.linalg.inv(matrix)
    print("The inverse of the matrix is:")
    print(inverse_matrix)
```

saddle points

```python
import sympy as sp

# Define variables
x, y = sp.symbols('x y')
a, b = sp.symbols('a b')

# Define the function
g = a**3 * b**2 - b

# Compute first-order partial derivatives
g_a = sp.diff(g, a)
g_ba = sp.diff(g_a, b)

# Evaluate at (2,3)
g_ba_value = g_ba.subs({a: 2, b: 3})

# Compute first-order partial derivatives for f
f = 4*x*y - x**4 - y**4 + 4
f_x = sp.diff(f, x)
f_y = sp.diff(f, y)

# Compute second-order partial derivatives for f
f_xx = sp.diff(f_x, x)
f_yy = sp.diff(f_y, y)
f_xy = sp.diff(f_x, y)

# Define Hessian determinant
D = f_xx * f_yy - f_xy**2

# Evaluate at (0,0)
d_value = D.subs({x: 0, y: 0})
f_xx_value = f_xx.subs({x: 0, y: 0})

# Determine critical point nature
if d_value > 0:
    if f_xx_value > 0:
        result = "Local Minimum"
    else:
        result = "Local Maximum"
elif d_value < 0:
    result = "Saddle Point"
else:
    result = "Inconclusive"

# Display results
print("Hessian Determinant at (0,0):", d_value)
print("f_xx at (0,0):", f_xx_value)
print("Nature of Critical Point (0,0):", result)
print("g_ba(2,3):", g_ba_value)
```
